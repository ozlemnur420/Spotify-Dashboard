## 3. Load Data

```python
import os

for root, dirs, files in os.walk("data"):
    level = root.replace("data", "").count(os.sep)
    indent = "    " * level
    print(f"{indent}{os.path.basename(root)}/")
    for file in files:
        print(f"{indent}    {file}")
```

```python
from pathlib import Path

def load_spotify_data(user_folder, user_name):
    folder = Path(user_folder) / "Spotify Extended Streaming History"

    audio_files = sorted(folder.glob("Streaming_History_Audio*.json"))
    video_files = sorted(folder.glob("Streaming_History_Video*.json"))

    audio_list = []
    video_list = []

    for file in audio_files:
        df = pd.read_json(file)
        df["user"] = user_name
        df["content_type"] = "Audio"
        audio_list.append(df)

    for file in video_files:
        df = pd.read_json(file)
        df["user"] = user_name
        df["content_type"] = "Video"
        video_list.append(df)

    audio_df = pd.concat(audio_list, ignore_index=True) if audio_list else pd.DataFrame()
    video_df = pd.concat(video_list, ignore_index=True) if video_list else pd.DataFrame()

    return audio_df, video_df
```

```python
ozlem_audio, ozlem_video = load_spotify_data("data/ozlem", "Özlem")
eren_audio, eren_video = load_spotify_data("data/eren", "Eren")

print("Özlem Audio :", ozlem_audio.shape)
print("Özlem Video :", ozlem_video.shape)
print("Eren Audio  :", eren_audio.shape)
print("Eren Video  :", eren_video.shape)
```

## 10. Music Taste Similarity

10.1 Common Artists

```python
ozlem_artists = set(
    spotify.loc[
        spotify["user"] == "Özlem",
        "master_metadata_album_artist_name"
    ].dropna()
)

eren_artists = set(
    spotify.loc[
        spotify["user"] == "Eren",
        "master_metadata_album_artist_name"
    ].dropna()
)

common_artists = ozlem_artists & eren_artists

print("Common artists:", len(common_artists))
```

10.2 Unique Artists

```python
only_ozlem = ozlem_artists - eren_artists
only_eren = eren_artists - ozlem_artists

print("Only Özlem:", len(only_ozlem))
print("Only Eren :", len(only_eren))
```

10.3 Music Taste Similarity Score

Similarity= Union/Common
	​


```python
similarity = (
    len(common_artists)
    /
    len(ozlem_artists | eren_artists)
)

print(f"Music Taste Similarity: {similarity:.2%}")
```

## 17. Listening Behavior Analysis

17.1 Skip Rate Analysis

```python
skip_analysis = (
    spotify
    .groupby("user")
    .agg(
        total_tracks=("skipped", "count"),
        skipped_tracks=("skipped", "sum")
    )
)

skip_analysis["skip_rate (%)"] = (
    skip_analysis["skipped_tracks"] /
    skip_analysis["total_tracks"] * 100
).round(2)

skip_analysis
```

```python
fig = px.bar(
    skip_analysis.reset_index(),
    x="user",
    y="skip_rate (%)",
    color="user",
    text="skip_rate (%)",
    title="Skip Rate by User",
    labels={
        "user": "User",
        "skip_rate (%)": "Skip Rate (%)"
    }
)

fig.show()
```

17.2 Shuffle Usage

```python
shuffle_analysis = (
    spotify
    .groupby("user")
    .agg(
        total_tracks=("shuffle", "count"),
        shuffle_used=("shuffle", "sum")
    )
)

shuffle_analysis["shuffle_rate (%)"] = (
    shuffle_analysis["shuffle_used"] /
    shuffle_analysis["total_tracks"] * 100
).round(2)

shuffle_analysis
```

```python
fig = px.bar(
    shuffle_analysis.reset_index(),
    x="user",
    y="shuffle_rate (%)",
    color="user",
    text="shuffle_rate (%)",
    title="Shuffle Usage Rate",
    labels={
        "user": "User",
        "shuffle_rate (%)": "Shuffle Usage (%)"
    }
)

fig.show()
```

17.3 Platform Usage Analysis

```python
platform_usage = (
    spotify
    .groupby(["user", "platform"])
    .size()
    .reset_index(name="play_count")
)

platform_usage
```

```python
fig = px.bar(
    platform_usage,
    x="platform",
    y="play_count",
    color="user",
    barmode="group",
    title="Platform Usage by User",
    labels={
        "platform": "Platform",
        "play_count": "Play Count",
        "user": "User"
    }
)

fig.show()
```

```python
spotify["platform_clean"] = spotify["platform"].str.lower()

spotify["platform_clean"] = (
    spotify["platform_clean"]
    .str.replace(r"android.*", "Android", regex=True)
    .str.replace(r"windows.*", "Windows", regex=True)
    .str.replace(r"web.*", "Web Player", regex=True)
    .str.replace(r"ios.*", "iOS", regex=True)
    .str.replace(r"mac.*", "macOS", regex=True)
)
```

```python
platform_usage = (
    spotify
    .groupby(["user", "platform_clean"])
    .size()
    .reset_index(name="play_count")
)

fig = px.bar(
    platform_usage,
    x="platform_clean",
    y="play_count",
    color="user",
    barmode="group",
    title="Platform Usage by User",
    labels={
        "platform_clean": "Platform",
        "play_count": "Play Count"
    }
)

fig.show()
```

17.4 Offline vs Online Listening

```python
offline_stats = (
    spotify
    .groupby(["user", "offline"])
    .size()
    .reset_index(name="play_count")
)

offline_stats["offline"] = offline_stats["offline"].map({
    True: "Offline",
    False: "Online"
})

offline_stats
```

```python
fig = px.bar(
    offline_stats,
    x="user",
    y="play_count",
    color="offline",
    barmode="stack",
    text="play_count",
    title="Offline vs Online Listening",
    labels={
        "user": "User",
        "play_count": "Play Count",
        "offline": "Listening Mode"
    }
)

fig.show()
```

17.5 Average Listening Duration

```python
avg_duration = (
    spotify
    .groupby("user")
    .agg(
        average_minutes=("ms_played", lambda x: x.mean() / 60000)
    )
    .round(2)
    .reset_index()
)

avg_duration
```

```python
fig = px.bar(
    avg_duration,
    x="user",
    y="average_minutes",
    color="user",
    text="average_minutes",
    title="Average Listening Time per Track",
    labels={
        "user": "User",
        "average_minutes": "Average Minutes"
    }
)

fig.show()
```
